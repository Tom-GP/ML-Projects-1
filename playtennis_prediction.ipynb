```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
playtennis_df = pd.DataFrame({
    'Day': ['D1', 'D2', 'D3', 'D4', 'D5', 'D6', 'D7', 'D8', 'D9', 'D10', 'D11', 'D12', 'D13', 'D14'],
    'outlook': ['sunny', 'sunny', 'overcast', 'rain', 'rain', 'rain', 'overcast', 'sunny', 'sunny', 'rain', 'sunny', 'overcast', 'overcast', 'rain'],
    'temperature': ['hot', 'hot', 'hot', 'mild', 'cool', 'cool', 'cool', 'mild', 'cool', 'mild', 'mild', 'mild', 'hot', 'mild'],
    'humidity': ['high', 'high', 'high', 'high', 'normal', 'normal', 'normal', 'high', 'normal', 'normal', 'normal', 'high', 'normal', 'high'],
    'wind': ['weak', 'strong', 'weak', 'weak', 'weak', 'strong', 'strong', 'weak', 'weak', 'weak', 'strong', 'strong', 'weak', 'strong'],
    'playtennis': ['no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'no', 'yes', 'yes', 'yes', 'yes', 'yes', 'no']
})

playtennis_df
```

```python
playtennis_df.drop(['Day'], axis=1, inplace=True)
playtennis_df
```

```python
le = LabelEncoder()
```

```python
playtennis_df['outlook_n'] = le.fit_transform(playtennis_df.outlook)
playtennis_df['temperature_n'] = le.fit_transform(playtennis_df.temperature)
playtennis_df['humidity_n'] = le.fit_transform(playtennis_df.humidity)
playtennis_df['wind_n'] = le.fit_transform(playtennis_df.wind)
```

```python
playtennis_df
```

```python
playtennis_df.drop(['outlook', 'temperature',	'humidity',	'wind'], axis=1, inplace=True)
```

```python
X = playtennis_df.drop('playtennis', axis=1)
Y = playtennis_df['playtennis']
```

```python
X
```

```python
Y
```

```python
model = DecisionTreeClassifier().fit(X, Y)
```

```python
model.score(X, Y)
```
