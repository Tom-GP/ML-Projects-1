```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('area_price.csv', header=1)
df = df[['area', 'price']].dropna()
df
```

```python
sns.scatterplot(data=df, x='area', y='price', color='red', marker='+')
plt.show()
```

```python
df.plot(kind='bar', x='area')
plt.xticks(rotation=30)
plt.ylabel('price')
plt.title('Area_Price Chart')
plt.show()
```

```python
model = LinearRegression()
model.fit(df[['area']], df.price)
```

```python
model.predict([[4000]])
```

```python
m = model.coef_
m
```

```python
b = model.intercept_
b
```

```python
sns.scatterplot(data=df, x='area', y='price', color='red')
plt.plot(df['area'], model.predict(df[['area']]))
plt.show()
```
