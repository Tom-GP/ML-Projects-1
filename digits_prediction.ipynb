```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
df = pd.DataFrame({
    'temperature': [85, 60, 80, 72, 68, 74, 69, 75, 83],
    'humidity': [85, 70, 90, 95, 80, 73, 70, 85, 78],
    'play': ['no', 'yes', 'no', 'no', 'yes', 'yes', 'yes', 'no', 'no']
})
df
```

```python
df_yes = df[df['play'] == 'yes']
df_no = df[df['play'] == 'no']
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(df_yes, x='temperature', y='humidity', marker='o', color='blue')
sns.scatterplot(df_no, x='temperature', y='humidity', marker='x', color='red')
plt.show()
```

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

```python
digits = load_digits()
dir(digits)
```

```python
df3 = pd.DataFrame(digits.data, columns=digits.feature_names)
df3.head(6)
```

```python
df3['target'] = digits.target
df3.sample(5)
```

```python
df3.info()
```

```python
X = df3.drop('target', axis=1)
X
```

```python
Y = df3.target
Y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
model = SVC(C=3.36).fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```
