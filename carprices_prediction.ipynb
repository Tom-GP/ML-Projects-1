```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
df = pd.read_csv('carprices.csv')
df
```

```python
le = LabelEncoder()
dfle = df.copy()
dfle['Car Model'] = le.fit_transform(dfle['Car Model'])
dfle
```

```python
ohe = OneHotEncoder()
dfhe = ohe.fit_transform(df[['Car Model']]).toarray()
dfhe
```

```python
dfhe = pd.DataFrame(dfhe, columns=ohe.get_feature_names_out(['Car Model']))
dfhe
```

```python
dfhe_merged = pd.concat([df.drop(['Car Model'], axis=1), dfhe], axis=1)
dfhe_merged
```

```python
x = dfhe_merged.drop(['Sell Price($)', 'Car Model_Mercedez Benz C class'], axis=1)
x
```

```python
y = dfhe_merged['Sell Price($)']
y
```

```python
model = LinearRegression().fit(x, y)
```

```python
model.predict([[45000,4,0,0]])
```

```python
model.predict([[86000, 7, 0, 1]])
```
