```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('canada_per_capita_income.csv')
df.head(5)
```

```python
sns.scatterplot(data=df, x='year', y='per capita income (US$)', color='red')
plt.show()
```

```python
model = LinearRegression().fit(df[['year']], df['per capita income (US$)'])
```

```python
model.predict([[2020]])
```

```python
model.coef_ # m
```

```python
model.intercept_ # b
```

```python
# y = mx + b
y = 828.46507522*2020+(-1632210.7578554575)
y
```

```python
sns.scatterplot(data=df, x='year', y='per capita income (US$)', color='red')
plt.plot(df.year, model.predict(df[['year']]))
plt.show()
```
