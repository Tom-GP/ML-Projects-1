```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso, Ridge
```

```python
df = pd.read_csv('Melbourne_housing_FULL.csv')
df.head()
```

```python
df.isna().sum()
```

```python
useful_cols = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount',
               'Distance', 'CouncilArea', 'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price']
```

```python
df2 = df[useful_cols]
df2.sample(5)
```

```python
df2.shape
```

```python
df2.isna().sum()
```

```python
cols_fill_zero = ['Propertycount', 'Distance', 'Bedroom2', 'Bathroom', 'Car']
df2[cols_fill_zero] = df2[cols_fill_zero].fillna(0)
df2.isna().sum()
```

```python
df2['Landsize'] = df2['Landsize'].fillna(df2['Landsize'].mean())
df2['BuildingArea'] = df2['BuildingArea'].fillna(df2['BuildingArea'].mean())
df2.isna().sum()
```

```python
df2.dropna(inplace=True)
```

```python
df2.isna().any()
```

```python
df2.info()
```

```python
df2.select_dtypes(include=['object']).columns
```

```python
df2 = pd.get_dummies(df2, drop_first=True).astype(int)
df2.info()
```

```python
X = df2.drop('Price', axis='columns')
y = df2.Price
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
model1 = LinearRegression().fit(X_train, y_train)
model2 = Lasso().fit(X_train, y_train)
model3 = Ridge().fit(X_train, y_train)
```

```python
model1.score(X_train, y_train)
```

```python
model1.score(X_test, y_test)
```

```python
model2.score(X_train, y_train)
```

```python
model2.score(X_test, y_test)
```

```python
model3.score(X_train, y_train)
```

```python
model3.score(X_test, y_test)
```
